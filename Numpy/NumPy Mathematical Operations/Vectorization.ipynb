```python
# Vectorization means performing operations on entire arrays without explicitly writing Python loops.
scores = [60, 70, 80, 90, 50]

new_scores = []

for score in scores:
    new_scores.append(score + 10)

print(new_scores)
```

```python
import numpy as np

scores = np.array([60, 70, 80, 90, 50])

new_scores = scores + 10

print(new_scores)
```
